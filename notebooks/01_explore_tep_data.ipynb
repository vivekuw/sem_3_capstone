# 01. Tennessee Eastman Process (TEP) - Exploratory Data Analysis
This notebook analyzes the baseline steady-state variables, sensor correlation matrices, and time-series distributions.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from ingestion.load_tep_dataset import generate_synthetic_benchmark

df_normal = generate_synthetic_benchmark(hours=3, fault_type="NORMAL")
print("Dataset shape:", df_normal.shape)
df_normal.head()
```

```python
# Plot correlation heatmap
plt.figure(figsize=(14, 10))
numeric_cols = [c for c in df_normal.columns if c != 'time']
corr = df_normal[numeric_cols].corr()
sns.heatmap(corr, cmap="vlag", center=0)
plt.title("Correlation Matrix across TEP Sensors")
plt.show()
```
